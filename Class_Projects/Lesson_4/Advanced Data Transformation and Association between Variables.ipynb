```python
import matplotlib.pyplot as plt
import pandas as pd 
import seaborn as sb
import numpy as np
```

```python
data=pd.read_csv('Titanic_Dataset.csv')
```

```python
data.head(5)
```

```python
minimum_age=data['Age'].min()
print("Minimum Age:", minimum_age)
```

```python
max_age=data['Age'].max()
print("Maximum Age:",max_age)
```

```python
bins=[0,15,30,45,60,65,75]
data['Binned_age']=pd.cut(data['Age'],bins)
print(data[['Binned_age','Age']].head())
age_labels = ['Young', 'Young - Adult', 'Middle Aged', 'Middle-Older Age', 'Senior','Very_Senior']
```

```python
data['Binned_age']=pd.cut(data['Age'],bins,labels=age_labels)
```

```python
data['Binned_age'].value_counts().plot(kind='bar')
plt.title('Dance Class Age Distributuon:')
plt.xlabel('Ages')
plt.ylabel('Counts')
```
