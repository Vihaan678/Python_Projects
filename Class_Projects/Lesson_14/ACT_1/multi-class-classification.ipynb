```python
# Importing necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn import datasets

# Custom Multi-Class Logistic Regression Class
class MultiClassLogisticRegression:
    """
    A wrapper class for multi-class logistic regression using One-vs-Rest strategy
    """
    def _init_(self, max_iter=1000):
        # Initialize the LogisticRegression model with multi_class parameter
        self.model = LogisticRegression(max_iter=max_iter, multi_class='ovr')
    
    def fit(self, X, y):
        """
        Fit the logistic regression model to the training data
        """
        self.model.fit(X, y)
        return self
    
    def predict_classes(self, X):
        """
        Predict class labels for samples in X
        """
        return self.model.predict(X)
    
    def score(self, X, y):
        """
        Return the mean accuracy on the given test data and labels
        """
        return self.model.score(X, y)

# Load the Iris dataset
iris = datasets.load_iris() # 150 samples and 4 features

# Select only the first two features (Sepal length and Sepal width)
X = iris.data[:, :2]  # Shape: (150, 2)

# Target variable (species: 0=Setosa, 1=Versicolor, 2=Virginica)
Y = iris.target

# Create an instance of our Multi-Class Logistic Regression
logreg = MultiClassLogisticRegression()

# Train the model on the data
logreg.fit(X, Y)

# Create a mesh grid for plotting decision boundaries
x_min, x_max = X[:, 0].min() - .5, X[:, 0].max() + .5
y_min, y_max = X[:, 1].min() - .5, X[:, 1].max() + .5
h = .02  # Step size in the mesh

# Generate grid points
xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))

# Predict class for each point in the mesh
Z = logreg.predict_classes(np.c_[xx.ravel(), yy.ravel()])

# Reshape predictions to match the mesh grid
Z = Z.reshape(xx.shape)

# Create the plot
plt.figure(1, figsize=(8, 6))

# Plot decision boundaries as colored regions
plt.pcolormesh(xx, yy, Z, cmap=plt.cm.Paired, alpha=0.8)

# Plot the training points
plt.scatter(X[:, 0], X[:, 1], c=Y, edgecolors='k', cmap=plt.cm.Paired, s=50)

# Add labels
plt.xlabel('Sepal length (cm)', fontsize=12)
plt.ylabel('Sepal width (cm)', fontsize=12)
plt.title('Multi-Class Logistic Regression Decision Boundaries', fontsize=14)

# Set axis limits
plt.xlim(xx.min(), xx.max())
plt.ylim(yy.min(), yy.max())

# Add a legend
plt.colorbar(label='Class')

# Display the plot
plt.tight_layout()
plt.show()

# Calculate and print the accuracy score
accuracy = logreg.score(X, Y)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
print(f"Number of samples: {len(X)}")
print(f"Number of classes: {len(np.unique(Y))}")
```
