```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('Weather Dataset.csv')
```

```python
data.head(5)
```

```python
data.isnull().sum()
```

```python
mean_temp=np.mean(data['Temperature (C)'])
print('mean temperature:',mean_temp)
```

```python
var=np.var(data['Temperature (C)'])
print('Variance of Temperature:',var)
```

```python
standard=np.std(data['Temperature (C)'])
print('Standard Variance of Temperature:',standard)
```

```python
for i in range(1,13):
    month=data.loc[data["month"]==i]["Temperature (C)"]
    print('month:'+ str(i))
    print('Mean Temperature:'+ str(np.mean(month)))
    print('Standard deviation:'+ str(np.std(month)))
```
